# Генерация учебных датасетов для Spark

Ноутбук создает синтетические parquet-датасеты под типовые Spark-сценарии: большой факт, маленький справочник, партиционированные данные и skew. Размер регулируется параметром `SCALE`.

```python
import sys
sys.path.append("/opt/workspace/scripts")

from spark_lab import make_spark, generate_datasets
```

```python
spark = make_spark(
    app_name="generate_spark_datasets",
    executor_memory="1g",
    executor_cores=1,
    cores_max=2,
    shuffle_partitions=16,
)

print("Spark UI:", spark.sparkContext.uiWebUrl)
```

## Размер данных

`SCALE=1` генерирует примерно 1 млн строк факта. Для заметной разницы между профилями ресурсов можно поставить `SCALE=3` или `SCALE=5`, если машине хватает RAM/диска.

```python
SCALE = 1
paths = generate_datasets(spark, base_path="data/lab", scale=SCALE)
paths
```

```python
for name, path in paths.items():
    if isinstance(path, str):
        print(name, path, spark.read.parquet(path).count())
```
